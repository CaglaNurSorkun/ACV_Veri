```python
numeric_cols = df.select_dtypes(include=['int64','float64']).columns
numeric_cols
```

```python
(df[numeric_cols] < 0).sum()
```

```python
Q1 = df["t0"].quantile(0.25)
Q3 = df["t0"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["t0"] < lower) | (df["t0"] > upper)]
```

```python
len(outliers)
```

```python
outlier_dict = {}

for col in numeric_cols:
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    
    outlier_dict[col] = len(outliers)

outlier_dict
```

IQR yöntemi kullanılarak aykırı değerler tespit edilmiştir. 
Enerji tüketim verilerinde bazı aykırı değerler gerçek operasyonel durumları temsil edebileceğinden bu değerler tamamen silinmemiş, analiz sırasında ayrıca değerlendirilmiştir.

# Ortalama Akım:
-Akım yüksek ama tüketim düşük → kaçak kullanım
-Akım düşük ama tüketim yüksek → sayaç hatası

```python
df["avg_current"] = (df["l1"] + df["l2"] + df["l3"]) / 3
```

# Ortalama Gerilim:
-Düşük voltaj → şebeke problemi

```python
df["avg_voltage"] = (df["v1"] + df["v2"] + df["v3"]) / 3
```

# Faz Dengesizliği
Fazlar çok farklıysa + yüksek akım varsa
-ekipman arızası / yük dengesizliği

```python
df["phase_imbalance"] = df[["l1","l2","l3"]].max(axis=1) - df[["l1","l2","l3"]].min(axis=1)
```

# Akım-Rüketim Tutarsızlığı
Dünkü tüketim ile bugünkü tüketim farkı ama bunu her abone için ayrı hesaplamak gerekir.
-Bir gün aşırı artış → kaçak kullanım
-Uzun süre sabit → sayaç hatası

```python
df = df.sort_values(["tesisat_no_id","load_profile_date"])

df["consumption_diff"] = df.groupby("tesisat_no_id")["t0"].diff()
```

# Gerilim Sapma Oranı
Gerilimin ortalamadan ne kadar saptığını ölçer.
-Yüksek sapma → şebeke problemi

```python
df["voltage_deviation"] = (
    df[["v1","v2","v3"]].max(axis=1) - df[["v1","v2","v3"]].min(axis=1)
) / df["avg_voltage"]
```

# Saat Dilimi (Gece / Gündüz / Mesai)
-Gece yüksek tüketim → kaçak kullanım

```python
df["hour"] = df["load_profile_date"].dt.hour
```

```python
def time_period(hour):
    if 0 <= hour < 6:
        return "night"
    elif 6 <= hour < 18:
        return "day"
    else:
        return "evening"

df["time_period"] = df["hour"].apply(time_period)
```
